```python
import numpy as np
import pandas as pd
```

```python
# Retrieve the amount paid > 1500 and payment method via GCash and PayMaya
# Sort the amount paid from highest to lowest
# Calculate the amount paid (total, average, minimum, and maximum)
# Group the total amount paid by payment method
# Write at least three findings in plain language

payments = pd.read_csv('payments.csv')
```

```python
payments
```

```python
# Retrieve the amount paid > 1500 and payment method via GCash and PayMaya
# Sort the amount paid from highest to lowest

payments_sorted = payments[
    (pd.to_numeric(payments['amount_paid'], errors = 'coerce') > 1500) & 
    ((payments['payment_method'] == 'GCash') | 
    (payments['payment_method'] == 'PayMaya'))
].sort_values('amount_paid', ascending = False)

# Run

payments_sorted
```

```python
# Calculate the amount paid (total, average, minimum, and maximum)
# Group the total amount paid by payment method

payments_sorted['amount_paid'] = pd.to_numeric(
    payments_sorted['amount_paid'],
    errors = 'coerce'
)

payments_aggregation = (
    payments_sorted.groupby('payment_method')
        .agg(
            amount_paid_sum = ('amount_paid', 'sum'),
            amount_paid_mean = ('amount_paid', 'mean'),
            amount_paid_min = ('amount_paid', 'min'),
            amount_paid_max = ('amount_paid', 'max')
        )
        .reset_index()
)

# Run

payments_aggregation
```

```python
# Write at least three findings in plain language

amount_paid_sum_GCash = payments_aggregation.iloc[0]['amount_paid_sum']
amount_paid_sum_PayMaya = payments_aggregation.iloc[1]['amount_paid_sum']

amount_paid_mean_GCash = payments_aggregation.iloc[0]['amount_paid_mean']
amount_paid_mean_PayMaya = payments_aggregation.iloc[1]['amount_paid_mean']

amount_paid_min_GCash = payments_aggregation.iloc[0]['amount_paid_min']
amount_paid_min_PayMaya = payments_aggregation.iloc[1]['amount_paid_min']

amount_paid_max_GCash = payments_aggregation.iloc[0]['amount_paid_max']
amount_paid_max_PayMaya = payments_aggregation.iloc[1]['amount_paid_max']
```

```python
print(f"In terms of the amount paid exceeding PHP 1,500.00:", end = "\n")

print(f"â¢ PayMaya registered a sum amounting to PHP {amount_paid_sum_PayMaya:,.2f} ââœ exceeding that of GCash amounting to PHP {amount_paid_sum_GCash:,.2f}.", end = "\n")

print(f"â¢ GCash registered a mean amounting to PHP {amount_paid_mean_GCash:,.2f} ââœ exceeding that of PayMaya amounting to PHP {amount_paid_mean_PayMaya:,.2f}.", end = "\n")

print(f"â¢ PayMaya received payments ranging from PHP {amount_paid_min_PayMaya:,.2f} to {amount_paid_max_PayMaya:,.2f} ââœ varying more than those of GCash ranging from PHP {amount_paid_min_GCash:,.2f} to {amount_paid_max_GCash:,.2f}.")
```

```python
def clean_currency(series):
    return pd.to_numeric(
        series.astype("string")
              .str.replace(
                  "₱", "", regex=False)
              .str.replace(
                  ",", "", regex=False)
              .str.strip(),
        errors="coerce"
    )
```

```python
items_analysis = items_clean.copy()

items_analysis["purchase_price_num"] = (
    clean_currency(
        items_analysis["unit_price_at_purchase"]
    )
)

items_analysis = items_analysis[
    (items_analysis["quantity"] > 0)
    & items_analysis["purchase_price_num"].notna()
]

items_analysis = items_analysis.merge(
    products_clean[["product_id", "unit_cost",
              "category_clean"]],
    on="product_id", how="left"
)

items_analysis["line_revenue"] = (
    items_analysis["quantity"]
    * items_analysis["purchase_price_num"]
)

items_analysis["line_cost"] = (
    items_analysis["quantity"]
    * items_analysis["unit_cost"]
)
```

```python
items_analysis
```

```python
sales_analysis = items_analysis.merge(
    orders_clean[[
        "order_id", "sales_channel", "order_status"
    ]],
    on="order_id", how="left"
)

channel_summary = (
    sales_analysis
    .groupby("sales_channel")
    .agg(
        gross_line_revenue=("line_revenue", "sum"),
        average_line_revenue=("line_revenue", "mean"),
        line_items=("order_item_id", "count"),
        orders=("order_id", "nunique")
    )
    .sort_values(
        "gross_line_revenue", ascending=False
    )
)

print(channel_summary)
```
